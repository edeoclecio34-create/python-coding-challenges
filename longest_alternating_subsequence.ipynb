```python
from typing import List

def alt_subsequence_best(X: List[int]) -> int:
    """
    Calcula o comprimento da maior subsequência alternada.
    """
    if not X:
        return 0

    up = 1
    down = 1

    for i in range(1, len(X)):
        if X[i] > X[i-1]:
            up = down + 1
        elif X[i] < X[i-1]:
            down = up + 1

    return max(up, down)

# --- SUÍTE DE TESTES AUTOMATIZADOS ---
def executar_testes():
    casos_de_teste = [
        {"input": [1, 7, 4, 9, 2, 5], "esperado": 6},
        {"input": [1, 17, 5, 10, 13, 15, 10, 5, 16, 8], "esperado": 7},
        {"input": [1, 2, 3, 4, 5, 6, 7, 8, 9], "esperado": 2},
        {"input": [], "esperado": 0},
        {"input": [5], "esperado": 1}
    ]

    print("🧪 Executando Testes Unitários...\n")
    for idx, teste in enumerate(casos_de_teste, 1):
        resultado = alt_subsequence_best(teste["input"])
        status = "✅ PASSOU" if resultado == teste["esperado"] else f"❌ FALHOU (Obtido: {resultado}, Esperado: {teste['esperado']})"
        print(f"Teste {idx}: Entrada = {teste['input']} | Resultado = {resultado} | {status}")

# Executa a suíte de testes
executar_testes()
```
